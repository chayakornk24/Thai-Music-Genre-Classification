### 1. Describe your project objective

The objective of this project is to classify Thai music into **four genres**:

- Luk Thung
- Pop
- Rock
- Hip-Hop

This project is a **supervised multiclass classification** problem.
The models use **61 audio features** extracted from 30-second music clips
as input.

The project aims to:

1. Develop models to classify Thai music genres.
2. Compare the performance of different models.
3. Analyze classification errors, especially between genres with similar characteristics.

### 5. Detail of your model implementation

### K-Nearest Neighbors (KNN)

The KNN model was implemented using `KNeighborsClassifier` from Scikit-learn.
It classifies each music clip based on the labels of its nearest training samples.

- **Input:** 61 scaled audio features
- **Output:** One of four genres: Luk Thung, Pop, Rock, or Hip-hop
- **Feature scaling:** StandardScaler fitted on the training set
- **Distance metric:** Euclidean distance
- **Neighbor weights:** Uniform

### Selecting K

Values of K from 1 to 29 were evaluated using the validation set.
K = 20 and K = 23 both achieved the lowest validation error of 25%.
K = 20 was selected for the final KNN model.

The selected model was then evaluated on the test set.

---

### Neural Network

The neural network models were implemented using Keras Sequential.
Each model takes 61 scaled audio features as input and predicts one of
four music genres: Luk Thung, Pop, Rock, or Hip-hop.

Three model configurations were compared:

| Model | Dense layers | Dropout |
|---|---|---:|
| Model 1 | 64 → 32 → 4 | 0.2 |
| Model 2 | 32 → 16 → 4 | 0.2 |
| Model 3 | 64 → 32 → 4 | 0.4 |

Each hidden Dense layer uses the ReLU activation function, followed by
a Dropout layer. The output layer uses Softmax to produce a score for
each of the four genres.

### Training configuration

- **Optimizer:** Adam
- **Loss function:** Sparse categorical cross-entropy
- **Maximum training epochs:** 100
- **Early stopping monitor:** Validation loss
- **Patience:** 10 epochs
- **Restore best weights:** Enabled

### Model selection

The three models were compared using validation loss and validation
accuracy. Model 3 had the lowest validation loss and the highest
validation accuracy among the three models, so it was selected for
evaluation on the test set.

### 6.3 Top-2 Accuracy

Some songs may contain characteristics of more than one genre,
while each clip in the dataset has only one label. Therefore, we
also evaluated the Neural Network using **Top-2 accuracy**.

A prediction is counted as correct under Top-2 accuracy when the
dataset label appears among the model's two highest-scoring genres.

| Metric | Accuracy | Correct clips |
|:--|--:|--:|
| Top-1 accuracy | 66.84% | 127 / 190 |
| Top-2 accuracy | 88.42% | 168 / 190 |

Among the **63 clips** incorrectly classified by Top-1, the correct
label appeared in second place for **41 clips**.

For example, the song *เดาไม่เก่ง* by Three Man Down was labeled
as Rock. The model predicted Pop as its first choice and Rock as
its second choice. Therefore, it was incorrect under Top-1 but
correct under Top-2.

Because the model predicts from only four classes, Top-2 includes
half of the available classes. It should therefore be considered
alongside Top-1 accuracy, rather than used as a replacement for it.

### 6.4 Comparison with Existing Studies

#### Study 1: Pereira et al. (2019)

Pereira et al. studied music genre classification using the Brazilian
Music Dataset, which contains **8,994 songs across six genres**. They
evaluated several classifiers and audio features using five-fold
cross-validation with an artist filter.

One of their experiments used **MFCC features with KNN (K = 10)** and
achieved an accuracy of **39.53%**.

Our project also used KNN and included MFCC among its audio features.
However, the datasets, genres, features, and evaluation methods were
different, so the accuracy values cannot be directly compared.

#### Study 2: Kamtue et al. (2019)

Kamtue et al. studied Thai music classification using a dataset of
**10,547 songs**. Their task was to classify songs as **Luk Thung or
non-Luk Thung**.

The study used Neural Networks with audio and lyrics. Its combined
model achieved a test **F1-score of 0.8561**.

Our project also includes Luk Thung music, but classifies clips into
four genres using audio features. The results cannot be directly
compared because the studies use different datasets, classification
tasks, inputs, and evaluation metrics.

| Study | Dataset and task | Method | Reported result |
|:--|:--|:--|:--|
| Pereira et al. (2019) | Brazilian music; 6 genres | MFCC + KNN | 39.53% accuracy |
| Kamtue et al. (2019) | Thai music; Luk Thung vs non-Luk Thung | Neural Network using audio and lyrics | 0.8561 F1-score |
| **Our project** | Thai music; 4 genres | Neural Network using audio features | 66.84% test accuracy |

The two external studies provide context for our approach. Their
reported results should not be used to directly rank the models,
because the experimental settings are different.

### 7. Discussion and Conclusion

### 7.1 Discussion

The Neural Network achieved a higher test accuracy than KNN:
66.84% compared with 63.16%. However, the class-level results show
that each model performed differently across genres. KNN achieved
higher recall for Luk Thung, while the Neural Network achieved
higher recall for Pop, Rock, and Hip-hop.

Both models had difficulty classifying Pop and Rock. In the Neural
Network results, 13 Pop clips were predicted as Luk Thung, and
13 Rock clips were predicted as Pop. One possible reason is that
some songs share musical characteristics across genres. However,
the confusion matrix alone cannot confirm the cause of these errors.

For example, *เดาไม่เก่ง* by Three Man Down was labeled as Rock,
but the Neural Network predicted Pop as its first choice and Rock
as its second choice. This example shows that the dataset label
was considered by the model, even though it was not the
highest-scoring prediction.

The Neural Network achieved 88.42% Top-2 accuracy, compared with
66.84% Top-1 accuracy. Among the 63 Top-1 errors, the correct
label appeared in second place for 41 clips. However, Top-2 covers
two out of the four available classes, so it is an additional
measure and should be interpreted alongside Top-1 accuracy.

### 7.2 Limitations

- The dataset contains four genres, and each clip has only one
  label, although some songs may have characteristics of more
  than one genre.
- The results are based on the selected test set. Performance may
  differ on other songs or datasets.
- The models use audio features only. They do not use lyrics or
  other information that may also help describe a song.

### 7.3 Conclusion

This project compared KNN and Neural Network models for classifying
Thai music into Luk Thung, Pop, Rock, and Hip-hop.

Among the three Neural Network models, Model 3 achieved the best
validation result. On the test set, it achieved **66.84% accuracy**,
while KNN achieved **63.16% accuracy**.

The results show that the Neural Network performed better overall
on this test set, although both models continued to make errors
between some genres. The Top-2 results provide additional
information about the model's predictions, especially when the
dataset label is not its first choice.
