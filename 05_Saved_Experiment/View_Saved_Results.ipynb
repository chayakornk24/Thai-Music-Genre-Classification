### Additional Analysis: Top-2 Accuracy

Top-2 accuracy was not covered in the course. However, during error analysis,
we observed that some songs may contain characteristics of more than one
genre, while each clip has only one label in the dataset.

To explore this issue, we used AI assistance to help write the code for
calculating Top-2 accuracy. This metric checks whether the true label is
among the model's two highest-scoring predictions.

```python
import numpy as np
from keras.models import load_model
```

```python
model = load_model("model_3.h5")
data = np.load("processed_data.npz")

x_test = data["x_test_scaled"]
y_true = data["y_test_number"]

if y_true.ndim > 1 and y_true.shape[1] > 1:
    y_true = np.argmax(y_true, axis=1)
else:
    y_true = y_true.reshape(-1)

y_prob = model.predict(x_test, verbose=0)
y_pred = np.argmax(y_prob, axis=1)

top1_correct = y_pred == y_true
top2_predictions = np.argsort(y_prob, axis=1)[:, -2:]
top2_correct = np.any(top2_predictions == y_true[:, None], axis=1)

top1_wrong = ~top1_correct
recovered = top1_wrong & top2_correct
```

```python
print(f"Top-1 Accuracy: {top1_correct.mean() * 100:.2f}%")
print(f"Top-2 Accuracy: {top2_correct.mean() * 100:.2f}%")
print(f"Top-1 Incorrect: {top1_wrong.sum()} clips")
print(f"Of these, correct label in Top-2: {recovered.sum()} clips")
```
